```python
from datasets import load_dataset
import pandas as pd

dataset = load_dataset("Inferencelab/RomanUrdu-NLP-Sentiment-Corpus")

print(dataset)
```

```python
data = dataset["train"]

print("Total records:", len(data))
print("\nColumn names:", data.column_names)

print("\nFirst 10 messages:")
for i in range(10):
    print("\nMessage:", data[i]["message"])
    print("Label:", data[i]["label"])
    print("-" * 50)
```

```python
df = data.to_pandas()

print("Sentiment distribution:")
print(df["label"].value_counts())

print("\nPercentage distribution:")
print((df["label"].value_counts(normalize=True) * 100).round(2))
```

```python
print("Missing messages:", df["message"].isna().sum())
print("Missing labels:", df["label"].isna().sum())
print("Duplicate messages:", df["message"].duplicated().sum())

print("\nDataset information:")
print(df.info())
```

```python
duplicates = df[df["message"].duplicated(keep=False)]

print("Total rows involved in duplicates:", len(duplicates))

print("\nExamples of duplicate messages:")
print(duplicates[["message", "label"]].head(20))

print("\nNumber of unique labels among duplicate messages:")
print(
    duplicates.groupby("message")["label"]
    .nunique()
    .value_counts()
)
```

```python
conflicts = (
    df.groupby("message")["label"]
    .nunique()
)

conflict_messages = conflicts[conflicts > 1].index

conflict_df = df[df["message"].isin(conflict_messages)]

print("Number of conflicting message groups:", len(conflict_messages))

print("\nTotal rows with conflicting labels:", len(conflict_df))

print("\nExamples:")
for message, group in conflict_df.groupby("message"):
    print("\nMessage:", message)
    print("Labels:", group["label"].tolist())

    if group.name == conflict_messages[4]:
        break
```

```python
conflicts = (
    df.groupby("message")["label"]
    .nunique()
)

conflict_messages = conflicts[conflicts > 1].index

conflict_df = df[df["message"].isin(conflict_messages)]

print("Conflicting message groups:", len(conflict_messages))
print("Total affected rows:", len(conflict_df))

for message, group in list(conflict_df.groupby("message"))[:10]:
    print("\nMessage:", repr(message))
    print("Labels:", group["label"].tolist())
    print("-" * 50)
```

```python
# Remove the one missing message
df_valid = df.dropna(subset=["message"]).copy()

# Identify messages with conflicting labels
label_counts = df_valid.groupby("message")["label"].nunique()

conflicting_messages = label_counts[label_counts > 1].index

# Remove all conflicting message groups
df_clean = df_valid[
    ~df_valid["message"].isin(conflicting_messages)
].copy()

# Keep one occurrence of each remaining message
df_clean = df_clean.drop_duplicates(subset=["message"])

print("Original records:", len(df))
print("After removing missing messages:", len(df_valid))
print("After removing conflicting groups:", len(df_clean))
print("\nRemaining sentiment distribution:")
print(df_clean["label"].value_counts())

print("\nRemaining duplicate messages:",
      df_clean["message"].duplicated().sum())
```

```python
for sentiment in ["Positive", "Neutral", "Negative"]:
    print("\n", "=" * 60)
    print("SENTIMENT:", sentiment)
    print("=" * 60)

    sample = df_clean[df_clean["label"] == sentiment].sample(
        n=10, random_state=42
    )

    for i, row in enumerate(sample.itertuples(), 1):
        print(f"\n{i}. {row.message}")
```

```python
# Select 20 random messages from each sentiment class

language_sample = pd.concat([
    df_clean[df_clean["label"] == sentiment].sample(
        n=20, random_state=42
    )
    for sentiment in ["Positive", "Neutral", "Negative"]
])

language_sample = language_sample.sample(
    frac=1, random_state=42
).reset_index(drop=True)

for i, row in language_sample.iterrows():
    print(f"\n{i+1}. {row['message']}")
    print("Sentiment:", row["label"])
```
